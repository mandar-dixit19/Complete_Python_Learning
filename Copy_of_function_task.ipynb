### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

````python
def unique_list(input_list):
    """
    Takes a list and returns a new list with unique elements.
    """
    return list(set(input_list))
````

````python
sample_list = [1, 2, 3, 3, 3, 3, 4, 5]
unique_elements = unique_list(sample_list)
print(f"Input: {sample_list}")
print(f"Output: {unique_elements}")
````

### Explanation for Problem-1:

1.  **`unique_list(input_list)` function**:
    *   This function takes one argument: `input_list`, which is the list you want to process.
    *   Inside the function, `set(input_list)` converts the input list into a `set`.
        *   A `set` is a built-in Python data type that stores only unique elements; any duplicate elements are automatically removed.
    *   `list(...)` then converts the `set` back into a `list`.
    *   The function returns this new list containing only unique elements.

2.  **Example Usage**:
    *   We define `sample_list` with duplicate numbers.
    *   We call `unique_list(sample_list)` to get the unique elements.
    *   The `print` statements show both the original input and the resulting list with unique elements, demonstrating the function's output as expected.

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

### Explanation for Problem-2:

To solve this problem, we need to perform three main steps:

1.  **Split the input string**: The input is a single string where words are separated by hyphens. We need to break this string into individual words.
2.  **Sort the words**: Once we have a list of individual words, we need to sort them alphabetically.
3.  **Join the sorted words**: Finally, we need to combine the sorted words back into a single hyphen-separated string.

````python
# Write code here
````

````python
def sort_hyphen_separated_words(hyphen_string):
    """
    Accepts a hyphen-separated sequence of words and returns the words
    in a hyphen-separated sequence after sorting them alphabetically.
    """
    # Step 1: Split the string into a list of words using the hyphen as a delimiter
    words = hyphen_string.split('-')

    # Step 2: Sort the list of words alphabetically
    words.sort()

    # Step 3: Join the sorted words back into a hyphen-separated string
    return '-'.join(words)

# Example usage:
input_string = "green-red-yellow-black-white"
sorted_string = sort_hyphen_separated_words(input_string)

print(f"Input: {input_string}")
print(f"Output: {sorted_string}")
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here
````

````python
def count_case_letters(sample_string):
    """
    Accepts a string and calculates the number of upper case and lower case letters.
    """
    upper_count = 0
    lower_count = 0
    for char in sample_string:
        if char.isupper():
            upper_count += 1
        elif char.islower():
            lower_count += 1
    return upper_count, lower_count

# Sample String provided in the problem description
sample_string = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'

upper_case_chars, lower_case_chars = count_case_letters(sample_string)

print(f"Sample String : '{sample_string}'")
print(f"No. of Upper case characters :  {upper_case_chars}")
print(f"No. of Lower case Characters :  {lower_case_chars}")
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here
````

````python
def is_perfect_number(number):
    """
    Checks if a given number is a perfect number.
    A perfect number is a positive integer that is equal to the sum of its proper positive divisors.
    """
    if number <= 0:
        return False # Perfect numbers are positive integers

    divisors_sum = 0
    for i in range(1, number):
        if number % i == 0:
            divisors_sum += i

    # A number is perfect if it is equal to the sum of its proper positive divisors
    return divisors_sum == number

# Example usage:
print(f"Is 6 a perfect number? {is_perfect_number(6)}")
print(f"Is 28 a perfect number? {is_perfect_number(28)}")
print(f"Is 496 a perfect number? {is_perfect_number(496)}")
print(f"Is 8128 a perfect number? {is_perfect_number(8128)}")
print(f"Is 12 a perfect number? {is_perfect_number(12)}")
print(f"Is 1 a perfect number? {is_perfect_number(1)}")
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
# Write code here
````

````python
def concatenate_dictionaries(*dicts):
    """
    Accepts any number of dictionaries as arguments and returns a new dictionary
    created by concatenating all the input dictionaries.
    """
    result_dict = {}
    for d in dicts:
        result_dict.update(d) # or result_dict = {**result_dict, **d}
    return result_dict

# Sample Dictionaries provided in the problem description
dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

# Concatenate the dictionaries
concatenated_dict = concatenate_dictionaries(dic1, dic2, dic3)

print(f"dic1={dic1}")
print(f"dic2={dic2}")
print(f"dic3={dic3}")
print(f"Expected Result : {concatenated_dict}")
````

`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
# Write code here
````

````python
from collections import Counter

def most_occurring_word(input_string):
    """
    Accepts a string as input and returns the word with the most occurrences
    and its count.
    """
    # Convert the string to lowercase to treat 'You' and 'you' as the same word
    # and split it into words.
    words = input_string.lower().split()

    # Use collections.Counter to count word occurrences
    word_counts = Counter(words)

    # Find the most common word and its count
    if word_counts:
        most_common = word_counts.most_common(1)[0]
        return most_common[0], most_common[1]
    else:
        return None, 0 # Return None if the string is empty or contains no words

# Example usage:
input_str = "hello how are you i am fine thank you"
word, count = most_occurring_word(input_str)

print(f"Input:\n{input_str}")
if word:
    print(f"Output\n{word} -> {count}")
else:
    print("Output\nNo words found in the input string.")
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here




````

````python
def create_histogram(numbers, bin_size=10):
    """
    Generates a histogram from a list of integers with a specified bin size.

    Args:
        numbers (list): A list of integers.
        bin_size (int): The size of each bin in the histogram.

    Returns:
        dict: A dictionary representing the histogram, where keys are bin ranges
              (e.g., '1-10', '11-20') and values are the counts of numbers
              falling into that bin.
    """
    histogram = {}
    for num in numbers:
        # Calculate the lower bound of the bin for the current number
        # For example, if num is 13 and bin_size is 10:
        # (13 - 1) // 10 * 10 + 1 = 12 // 10 * 10 + 1 = 1 * 10 + 1 = 11
        # So, the bin starts at 11.
        # This handles numbers like 10, 20, 30 correctly, placing them in their respective higher bin.
        bin_start = ((num - 1) // bin_size) * bin_size + 1
        bin_end = bin_start + bin_size - 1

        # Create the bin key string (e.g., '11-20')
        bin_key = f"{bin_start}-{bin_end}"

        # Increment the count for this bin
        histogram[bin_key] = histogram.get(bin_key, 0) + 1
    return histogram

# Sample Input:
input_list = [13, 42, 15, 37, 22, 39, 41, 50]

# Generate the histogram
histogram_output = create_histogram(input_list)

print(f"Input:\n{input_list}")
print(f"Output:\n{histogram_output}")
````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
# Write code here
````

````python
import math

def find_closest_coordinate(coordinates, query_point):
    """
    Accepts a list of 2D coordinates and a query point, and then finds
    the coordinate which is closest in terms of Euclidean distance from the query point.

    Args:
        coordinates (list): A list of tuples, where each tuple represents a 2D coordinate (x, y).
        query_point (tuple): A tuple representing the query 2D point (x, y).

    Returns:
        tuple: The coordinate from the list that is closest to the query point.
    """
    if not coordinates:
        return None

    closest_coord = None
    min_distance = float('inf') # Initialize with a very large distance

    qx, qy = query_point

    for coord in coordinates:
        cx, cy = coord
        # Calculate Euclidean distance
        distance = math.sqrt(math.pow(cx - qx, 2) + math.pow(cy - qy, 2))

        if distance < min_distance:
            min_distance = distance
            closest_coord = coord

    return closest_coord

# Sample Input:
list_of_coordinates = [(1, 1), (2, 2), (3, 3), (4, 4)]
query_point = (0, 0)

# Find the closest coordinate
nearest = find_closest_coordinate(list_of_coordinates, query_point)

print(f"List of Coordinates\n{list_of_coordinates}")
print(f"Query Point\n{query_point}")
print(f"Output\nNearest to {query_point} is {nearest}")
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
from collections import Counter
import re

def bag_of_words(list_of_strings):
    """
    Performs a Bag-of-Words operation on a list of strings.

    Args:
        list_of_strings (list): A list of strings (documents).

    Returns:
        tuple: A tuple containing:
               - vocabulary (list): A sorted list of unique words found across all documents.
               - bow_vectors (list): A list of dictionaries, where each dictionary represents
                 a document's bag-of-words vector (word counts).
    """
    all_words = []
    for text in list_of_strings:
        # Tokenize by splitting on non-alphanumeric characters and convert to lowercase
        words = re.findall(r'\b\w+\b', text.lower())
        all_words.extend(words)

    # Create a vocabulary of unique words
    vocabulary = sorted(list(set(all_words)))

    # Generate Bag-of-Words vectors for each document
    bow_vectors = []
    for text in list_of_strings:
        words = re.findall(r'\b\w+\b', text.lower())
        word_counts = Counter(words)

        # Create a vector for the current document based on the vocabulary
        vector = {word: word_counts.get(word, 0) for word in vocabulary}
        bow_vectors.append(vector)

    return vocabulary, bow_vectors

# Example Usage:
corpus = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"
]

vocabulary, bow_representation = bag_of_words(corpus)

print("Input Documents:")
for i, doc in enumerate(corpus):
    print(f"Document {i+1}: {doc}")

print("\nVocabulary:", vocabulary)

print("\nBag-of-Words Representation:")
for i, vec in enumerate(bow_representation):
    print(f"Document {i+1} Vector: {vec}")
````

###`Problem-12:`Write a Python program to create a list containing the power of said number in bases raised to the corresponding number in the index using Python map.
`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,2,9,64,625,-]
```


````python
# Write code here
````

````python
list1 = [1, 2, 3, 4, 5, 6]

# Use map with a lambda function and zip to pair each number with its index.
# The lambda function calculates `number ** index` for each pair.
# We convert the map object to a list.
result_list = list(map(lambda item: item[0] ** item[1], zip(list1, range(len(list1)))))

print(f"Input:\n{list1}")
print(f"Output:\n{result_list}")
````

### Explanation for Problem-12:

1.  **`list1 = [1, 2, 3, 4, 5, 6]`**: This is the given input list.

2.  **`range(len(list1))`**: This generates a sequence of numbers from `0` to `len(list1) - 1`, which represent the indices of the elements in `list1`. For `list1`, this will be `[0, 1, 2, 3, 4, 5]`.

3.  **`zip(list1, range(len(list1)))`**: The `zip` function pairs corresponding elements from `list1` and the generated index sequence. This creates an iterable of tuples: `[(1, 0), (2, 1), (3, 2), (4, 3), (5, 4), (6, 5)]`.

4.  **`lambda item: item[0] ** item[1]`**: This is an anonymous function (lambda function).
    *   It takes one argument, `item`, which will be each tuple generated by `zip` (e.g., `(1, 0)`, `(2, 1)`, etc.).
    *   `item[0]` refers to the number from `list1` (the base).
    *   `item[1]` refers to the index (the exponent).
    *   `item[0] ** item[1]` calculates the number raised to the power of its index.

5.  **`map(...)`**: The `map` function applies the `lambda` function to each tuple generated by `zip`. So, for each `(number, index)` pair, it computes `number ** index`.

6.  **`list(...)`**: Finally, `list()` converts the result of the `map` object into a list, giving us the desired output:
    *   `1**0 = 1`
    *   `2**1 = 2`
    *   `3**2 = 9`
    *   `4**3 = 64`
    *   `5**4 = 625`
    *   `6**5 = 7776`

    The output `[1, 2, 9, 64, 625, 7776]` matches the pattern indicated by the example output (ignoring the '-' which likely represents a truncated or omitted value).

###`Problem-13` Using filter() and list() functions and .lower() method filter all the vowels in a given string.



````python
# Write code here
````

````python
# Define the input string
input_string = "Hello World! This is a test string."

# Define a set of vowels for efficient lookup
vowels = {'a', 'e', 'i', 'o', 'u'}

# Use filter() to get only the vowels
# The lambda function converts each character to lowercase before checking if it's in the vowels set.
filtered_vowels = list(filter(lambda char: char.lower() in vowels, input_string))

print(f"Original String: '{input_string}'")
print(f"Filtered Vowels: {filtered_vowels}")
````

### Explanation for Problem-13:

1.  **`input_string = "Hello World! This is a test string."`**: This is the string from which we want to extract vowels.

2.  **`vowels = {'a', 'e', 'i', 'o', 'u'}`**: We create a `set` of lowercase vowels. Using a `set` for vowel checking is efficient because `in` operations on sets have an average time complexity of O(1).

3.  **`lambda char: char.lower() in vowels`**: This is a `lambda` (anonymous) function that takes a single character (`char`) as input:
    *   `char.lower()`: Converts the character to its lowercase equivalent. This ensures that both 'A' and 'a' are checked against the lowercase vowel set.
    *   `... in vowels`: Checks if the lowercase character is present in our `vowels` set. This expression returns `True` if it's a vowel, `False` otherwise.

4.  **`filter(lambda char: ..., input_string)`**: The `filter()` function applies the `lambda` function to each character in the `input_string`. It constructs an iterator that yields only those characters for which the `lambda` function returns `True` (i.e., only the vowels).

5.  **`list(...)`**: Finally, `list()` converts the iterator returned by `filter()` into a new list, which contains all the vowels found in the original string.

`Problem-14`: Use reduce to convert a 2D list to 1D

````python
# Write code here
````

````python
from functools import reduce

# Define a 2D list (list of lists)
list_2d = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]

# Use reduce to flatten the 2D list into a 1D list
# The lambda function takes two arguments: 'acc' (accumulator) and 'sublist'.
# It concatenates the accumulator with the current sublist.
list_1d = reduce(lambda acc, sublist: acc + sublist, list_2d)

print(f"Original 2D List: {list_2d}")
print(f"Flattened 1D List: {list_1d}")
````

### Explanation for Problem-14:

1.  **`from functools import reduce`**: We need to import the `reduce` function from the `functools` module, as it's not a built-in function.

2.  **`list_2d = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]`**: This is our sample 2D list, which is a list containing other lists (sublists).

3.  **`lambda acc, sublist: acc + sublist`**: This is a `lambda` (anonymous) function that defines the operation `reduce` will perform.
    *   `acc` (accumulator): This variable holds the result of the previous operation. Initially, it's the first element of the `list_2d` (i.e., `[1, 2, 3]`).
    *   `sublist`: This variable represents the current element from the `list_2d` that `reduce` is processing (e.g., `[4, 5]`, then `[6, 7, 8, 9]`).
    *   `acc + sublist`: This performs list concatenation. It joins the `acc` list with the `sublist`, effectively flattening them.

4.  **`reduce(lambda acc, sublist: acc + sublist, list_2d)`**: The `reduce` function applies the `lambda` function cumulatively to the items of `list_2d`, from left to right, so as to reduce the list to a single value.
    *   **Step 1:** `acc = [1, 2, 3]`, `sublist = [4, 5]`. Result: `[1, 2, 3, 4, 5]`
    *   **Step 2:** `acc = [1, 2, 3, 4, 5]`, `sublist = [6, 7, 8, 9]`. Result: `[1, 2, 3, 4, 5, 6, 7, 8, 9]`

5.  The final result of `reduce` is the completely flattened 1D list.
