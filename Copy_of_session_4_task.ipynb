###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
# Write code here
````

````python
list = [2,4,6,10,1]
result= []
for i in range(len(list)):
  current_num = list[i]
  sum = current_num
  for j in range(len(list)):
    if i != j and list[j] > current_num:
      sum += list[j]
  result.append(sum)
print(result)


````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
# Write code here
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]

# Convert to sets to find common elements efficiently
set1 = set(num1)
set2 = set(num2)

# Find the intersection of the two sets
common_elements = list(set1.intersection(set2))

# Sort the common elements in increasing order
common_elements.sort()

print(common_elements)
````

###`Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`
```
['1ac21', '23fg', '456', '098d','1','kls']
```

`Output:`
```
['456', '23fg', '1ac21', '1', 'kls', '098d']

```

````python
def get_numeric_product(s):
    product = 1
    found_digit = False
    for char in s:
        if char.isdigit():
            product *= int(char)
            found_digit = True
    return product if found_digit else 1

alphnum_list = ['1ac21', '23fg', '456', '098d','1','kls']

alphnum_list.sort(key=get_numeric_product, reverse=True) # Sort in descending order of product to match output example

print(alphnum_list)
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
# Code here
````

````python
input_list = ['campusxIs', 'bestFor', 'dataScientist']

output_list_alternative = []
for word in input_list:
    new_word = []
    for i, char in enumerate(word):
        if i > 0 and char.isupper() and word[i-1].islower():
            new_word.append(' ')
        new_word.append(char)
    output_list_alternative.append(''.join(new_word))

print(output_list_alternative)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

````python
# Write code here
````

````python
list1 = [1,2,3,4,5,1]
list2 = [2,3,5,7,8]

# Convert lists to sets to find the union (removes duplicates automatically)
set1 = set(list1)
set2 = set(list2)

# Perform the union operation
union_set = set1.union(set2)

# Convert the union set back to a list and sort it
result_list = sorted(list(union_set))


print(result_list)
````

### `Problem 14:` Write a list comprehension that can transpose a given matrix

matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

[1, 4, 7]<br>
[2, 5, 8]<br>
[3, 6, 9]<br>

````python
# Write code here
````

````python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

# Transpose the matrix using a list comprehension
transposed_matrix = [[row[i] for row in matrix] for i in range(len(matrix[0]))]

print(transposed_matrix)
````

### `Problem 15:` Write a list comprehension that can flatten a nested list

Input<br>
matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

Output:<br>
[1, 2, 3, 4, 5, 6, 7, 8, 9]

````python
# Write code here
````

````python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

# Flatten the matrix using a list comprehension
flattened_list = [item for sublist in matrix for item in sublist]

print(flattened_list)
````
