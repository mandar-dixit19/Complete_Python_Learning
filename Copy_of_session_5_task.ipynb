###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
print(f"The original list is: {test_list}")

# Create a dictionary to group elements by their first item
grouped_elements = {}
for tpl in test_list:
    first_element = tpl[0]
    print(first_element)
#     # If the first element is already a key, extend the list of remaining elements
    if first_element in grouped_elements:
        grouped_elements[first_element].extend(tpl[1:])
    # Otherwise, create a new entry with the first element and the rest of the tuple
    else:
        grouped_elements[first_element] = list(tpl[1:])

# Reconstruct the result list of tuples
result = []
for key, values in grouped_elements.items():
    # Combine the key with the sorted unique values
    result_tuple = (key,) + tuple(sorted(list(set(values))))
    result.append(result_tuple)

print(f"The joined tuples list is: {result}")
````

```
Set-2
Tuples-1
```

````python
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

tuple_count = 0
list_count = 0
set_count = 0

for item in list1:
    if isinstance(item, tuple):
        tuple_count += 1
    elif isinstance(item, list):
        list_count += 1
    elif isinstance(item, set):
        set_count += 1

print(f"List-{list_count}")
print(f"Set-{set_count}")
print(f"Tuples-{tuple_count}")
````

###`Q5`: Shortlist Students for a Job role
Ask user to input students record and store in tuples for each record. Then Ask user to input three things he wants in the candidate- Primary Skill, Higher Education, Year of Graduation.

Show every students record in form of tuples if matches all required criteria.

It is assumed that there will be only one primry skill.

If no such candidate found, print `No such candidate`

`Input:`
```
Enter No of records- 2
Enter Details of student-1
Enter Student name- Manohar
Enter Higher Education- B.Tech
Enter Primary Skill- Python
Enter Year of Graduation- 2022
Enter Details of student-2
Enter Student name- Ponian
Enter Higher Education- B.Sc.
Enter Primary Skill- C++
Enter Year of Graduation- 2020

Enter Job Role Requirement
Enter Skill- Python
Enter Higher Education- B.Tech
Enter Year of Graduation- 2022
```

`Output`
```
('Manohar', 'B.tech', 'Python', '2022')
```


````python
# write your code here
````

````python
num_records = int(input("Enter No of records- "))

students = []
for i in range(num_records):
    print(f"Enter Details of student-{i+1}")
    name = input("Enter Student name- ")
    education = input("Enter Higher Education- ")
    skill = input("Enter Primary Skill- ")
    year = int(input("Enter Year of Graduation- "))
    students.append((name, education, skill, year))

print("\nEnter Job Role Requirement")
required_skill = input("Enter Skill- ")
required_education = input("Enter Higher Education- ")
required_year = int(input("Enter Year of Graduation- "))

matching_students = []
for student in students:
    if (student[2] == required_skill and
        student[1] == required_education and
        student[3] == required_year):
        matching_students.append(student)

if matching_students:
    for student in matching_students:
        print(student)
else:
    print("No such candidate")
````

### Code Explanation:

1.  **Input Number of Records**: It first prompts the user to enter the total number of student records they want to input.
2.  **Collect Student Data**: It then iterates `num_records` times, each time asking for the student's name, higher education, primary skill, and year of graduation. Each student's data is stored as a tuple and added to a list called `students`.
3.  **Input Job Requirements**: After collecting all student data, it asks the user to input the required primary skill, higher education, and year of graduation for the job role.
4.  **Filter Matching Students**: It then iterates through the `students` list. For each student tuple, it checks if their primary skill, higher education, and year of graduation match the required criteria. If all criteria match, the student's tuple is added to `matching_students` list.
5.  **Display Results**: Finally, it checks if the `matching_students` list is not empty. If there are matching candidates, it prints each student's tuple. Otherwise, it prints "No such candidate".

### Solution for `Q2`

````python
# write your code here
````

````python
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"

# Define all possible vowels (lowercase and uppercase, as they are considered different)
vowels = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}

# Initialize an empty set to store unique vowels found in the string
unique_vowels_found = set()

# Iterate through each character in the string
for char in Str1:
    # Check if the character is in our defined set of vowels
    if char in vowels:
        # Add the vowel to the set of unique vowels found
        unique_vowels_found.add(char)

# The number of unique vowels is the length of the set
print(f"No of unique vowels-{len(unique_vowels_found)}")
````

### `Q4`: find union of n arrays.

**Example 1:**

Input:
```bash
[[1, 2, 2, 4, 3, 6],
 [5, 1, 3, 4],
 [9, 5, 7, 1],
 [2, 4, 1, 3]]
```

Output:

```bash
[1, 2, 3, 4, 5, 6, 7, 9]
```

````python
input_arrays = [
    [1, 2, 2, 4, 3, 6],
    [5, 1, 3, 4],
    [9, 5, 7, 1],
    [2, 4, 1, 3]
]

print(f"Input arrays: {input_arrays}")

# Initialize an empty set to store the union of all elements
union_set = set()

# Iterate through each array and add its elements to the union_set
for arr in input_arrays:
    union_set.update(arr)
    print(union_set)

# Convert the union set back to a sorted list
output_list = sorted(list(union_set))

print(f"Union of arrays: {output_list}")
````

````python
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}

print(f"Input dictionary: {test_dict}")

max_unique_count = -1
key_with_max_unique = None

for key, value_list in test_dict.items():
    unique_values = set(value_list)
    current_unique_count = len(unique_values)

    if current_unique_count > max_unique_count:
        max_unique_count = current_unique_count
        key_with_max_unique = key

print(f"Key with maximum unique values: {key_with_max_unique}")

# Example 2:
test_dict_2 = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}

print(f"\nInput dictionary (Example 2): {test_dict_2}")

max_unique_count_2 = -1
key_with_max_unique_2 = None

for key, value_list in test_dict_2.items():
    unique_values = set(value_list)
    current_unique_count = len(unique_values)

    if current_unique_count > max_unique_count_2:
        max_unique_count_2 = current_unique_count
        key_with_max_unique_2 = key

print(f"Key with maximum unique values (Example 2): {key_with_max_unique_2}")
````

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
# write your code here
````

````python
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]

print(f"Original list: {test_list}")
print(f"Keys list: {key_list}")

list_of_dicts = []
# Iterate through the test_list, taking 'len(key_list)' elements at a time
for i in range(0, len(test_list), len(key_list)):
    # Get the current chunk of values
    current_values = test_list[i:i + len(key_list)]
    # Create a dictionary by zipping key_list with current_values
    temp_dict = dict(zip(key_list, current_values))
    list_of_dicts.append(temp_dict)

print(f"List of dictionaries: {list_of_dicts}")

# Example 2:
test_list_2 = ["CampusX", 10]
key_list_2 = ["name", "id"]

print(f"\nOriginal list (Example 2): {test_list_2}")
print(f"Keys list (Example 2): {key_list_2}")

list_of_dicts_2 = []
for i in range(0, len(test_list_2), len(key_list_2)):
    current_values_2 = test_list_2[i:i + len(key_list_2)]
    temp_dict_2 = dict(zip(key_list_2, current_values_2))
    list_of_dicts_2.append(temp_dict_2)

print(f"List of dictionaries (Example 2): {list_of_dicts_2}")
````

Output:


````python
# write your code here
````

````python
test_dict = {'c': [3], 'b': [12, 10], 'a': [19, 4]}

print(f"Original dictionary: {test_dict}")

sorted_dict = {}
# Sort the dictionary items by key, then sort the values list for each key
for key in sorted(test_dict.keys()):
    sorted_dict[key] = sorted(test_dict[key])

print(f"Sorted dictionary: {sorted_dict}")

# Example 2:
test_dict_2 = {'c': [10, 34, 3]}

print(f"\nOriginal dictionary (Example 2): {test_dict_2}")

sorted_dict_2 = {}
for key in sorted(test_dict_2.keys()):
    sorted_dict_2[key] = sorted(test_dict_2[key])

print(f"Sorted dictionary (Example 2): {sorted_dict_2}")
````
