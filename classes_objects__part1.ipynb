```python
class Atm:

  # constructor is a function inside the classs  it is a special function => superpower => does ot need to execute it can be execute automatocally when object is created
  def __init__(self):
   self.pin = ""
   self.balance = 0
   print(id(self)) # the id of self this nd obj is same
   self.menu()


  def menu(self):
    user_input =  input("""
      hi how can i help you
      1.press 1 to create pin
      2.press 2 to change pin
      3.press 3 to check balance
      4.press 4 to withdrow
      5.press 5 to else
     """)

    if user_input == "1":
      # create pin
      self.create_pin()
      self.menu()
    elif user_input == "2":
      # change pin
      self.change_pin()
    elif user_input == "3":
      #check balan
      self.check_balance()
      self.menu()
    elif user_input == "4":
      #withdrow
       self.withdraw()
    else:
     exit()

  def create_pin(self):
    user_pin = input("enter your pin")
    self.pin = user_pin

    user_balance = int(input("enter your balance"))
    self.balance = user_balance

    print("pin created successfully")

  def change_pin(self):
     old_pin = input("enter your old pin")
     if old_pin == self.pin:
      new_pin = input("enter new pin")
      self.pin = new_pin
      print("pin change successfully")
      self.menu()
     else:
      print("you entered wrong pin")
      self.menu()

  def check_balance(self):
     user_pin = input("emter your pin")
     if user_pin == self.pin:
      print('your balance is ', self.balance)
     else:
      print("chal nikal yahase")

  def withdraw(self):
    user_pin = input("enter the pin")
    if user_pin == self.pin:
        amount = int(input("enter the amount"))
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("withdraw successfully balance is ", self.balance)
            self.menu()
        else:
            print("abe garib")
    else:
        print("sale chor")
    self.menu()










```

```python
# lets make a data type fraction
class fraction:

  # this is the parameterised constructor
  def __init__(self,x,y):
    self.num = x;
    self.den = y
  # here when we put obj inside the print the str code is execute
  def __str__(self):
    return "{}/{}".format(self.num, self.den)

  # let give the function to add the num and den
  def __add__(self, other):
    new_num = self.num*other.den + self.den*other.num
    new_den = self.den*other.den
    return "{}/{}".format(new_num, new_den)
    #26/20
  # substract
  def __sub__(self, other):
    new_num = self.num*other.den - self.den*other.num
    new_den = self.den*other.den
    return "{}/{}".format(new_num, new_den)
  #multipliction
  def __mul__(self, other):
    new_num = self.num*other.den * self.den*other.num
    new_den = self.den*other.den
    return "{}/{}".format(new_num, new_den)

  # divvision
  def __truediv__(self, other):
    new_num = self.num*other.den
    new_den = self.den*other.den
    return "{}/{}".format(new_num, new_den)

```

```python
fr1 = fraction(2,4)
fr2 = fraction(4,5)
print(fr1) #
print(fr2)

print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
```
