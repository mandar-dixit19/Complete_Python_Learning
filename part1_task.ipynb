### Q2:- Write a program that will convert celsius value to fahrenheit.

print(...): This is a standard Python function used to display output to the console.
'%.2f Celsius is: %0.2f Fahrenheit': This is a format string. It's a blueprint for how the output will look.
%.2f: This is a format specifier that tells Python to insert a floating-point number (decimal number) here, formatted to two decimal places. The . indicates precision, and 2 specifies two digits after the decimal point.
%0.2f: This is similar to %.2f. The 0 typically indicates zero-padding, but in the context of f (float) with precision, it usually behaves the same as %.2f for positive numbers, ensuring two decimal places.
Celsius is: and Fahrenheit: These are literal strings that will be printed as they are.
%(celsius, fahrenheit): This is the operator for old-style (or printf-style) string formatting in Python. It takes the format string on its left and a tuple of values on its right.
The values in the tuple (celsius, fahrenheit) are inserted into the format string from left to right, corresponding to the format specifiers (%.2f and %0.2f). So, the value of celsius goes into the first %.2f, and the value of fahrenheit goes into the second %0.2f.



````python
# Write your code here
celsius = float(input("Enter temperature in celsius: "))
fahrenheit = (celsius * 9/5) + 32
print('%.2f Celsius is: %0.2f Fahrenheit' %(celsius, fahrenheit))
print(f"the value in celcius is {celsius} and the value in fer {fahrenheit}")
# here the values at the

cel = float(input("enter the value in celcius"))
fer = (cel  * 9/5) + 32
print(f"the value in cel is {cel} and the value in fer is {fer}")
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("Enter total number of heads: "))
legs = int(input("Enter total number of legs: "))

# Let d be the number of dogs and c be the number of chickens.
# d + c = heads
# 4d + 2c = legs

# From the first equation, c = heads - d
# Substitute into the second equation:
# 4d + 2(heads - d) = legs
# 4d + 2*heads - 2d = legs
# 2d = legs - 2*heads
# d = (legs - 2*heads) / 2

dogs = (legs - 2 * heads) / 2
chickens = heads - dogs

# Check if the solution is valid (non-negative integers)
if dogs >= 0 and chickens >= 0 and dogs == int(dogs) and chickens == int(chickens):
    print(f"dogs -> {int(dogs)}")
    print(f"chicken -> {int(chickens)}")
else:
    print("Cannot find a valid solution with the given heads and legs.")
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a1 = float(input("Enter the first term of the arithmetic series: "))
a2 = float(input("Enter the second term of the arithmetic series: "))
n_term = int(input("Enter the value of N (the term number to find): "))

# Calculate the common difference
d = a2 - a1

# Calculate the Nth term using the formula: an = a1 + (n - 1) * d
an = a1 + (n_term - 1) * d

print(f"The {n_term}th term of the arithmetic series is: {an:.2f}")
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
import math

# Get input for the first fraction
numerator1 = int(input("Enter the numerator of the first fraction: "))
denominator1 = int(input("Enter the denominator of the first fraction: "))

# Get input for the second fraction
numerator2 = int(input("Enter the numerator of the second fraction: "))
denominator2 = int(input("Enter the denominator of the second fraction: "))

# Calculate the sum of the fractions
# Common denominator is denominator1 * denominator2
sum_numerator = (numerator1 * denominator2) + (numerator2 * denominator1)
sum_denominator = denominator1 * denominator2

# Simplify the resulting fraction using greatest common divisor (GCD)
gcd = math.gcd(sum_numerator, sum_denominator)

simplified_numerator = sum_numerator // gcd
simplified_denominator = sum_denominator // gcd

print(f"The sum of the fractions is: {simplified_numerator}/{simplified_denominator}")
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
import math

# Get dimensions of the milk tank
tank_height = float(input("Enter the height of the milk tank (cm): "))
tank_length = float(input("Enter the length of the milk tank (cm): "))
tank_breadth = float(input("Enter the breadth of the milk tank (cm): "))

# Get dimensions of the glass
glass_height = float(input("Enter the height of the glass (cm): "))
glass_radius = float(input("Enter the radius of the glass (cm): "))

# Calculate volume of the milk tank (cuboid)
tank_volume = tank_height * tank_length * tank_breadth

# Calculate volume of a single glass (cylinder)
glass_volume = math.pi * (glass_radius**2) * glass_height

# Calculate how many glasses can be obtained
# Use int() to get a whole number of glasses, as we can't fill a fraction of a glass
num_glasses = int(tank_volume / glass_volume)

print(f"The volume of the milk tank is: {tank_volume:.2f} cubic cm")
print(f"The volume of one glass is: {glass_volume:.2f} cubic cm")
print(f"Number of glasses of milk that can be obtained: {num_glasses}")
````
