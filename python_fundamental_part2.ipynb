### Sequence sum

1/1! + 2/2! + 3/3! + ...

```python
num_terms = int(input("Enter the number of terms for the series: "))

series_sum = 0
current_factorial = 1 # Initialize factorial for 0! or 1!, to be updated in the loop

for i in range(1, num_terms + 1):
    # Calculate the factorial for the current 'i'
    current_factorial *= i # This updates current_factorial to be i!

    term = i / current_factorial
    series_sum += term

print(f"The sum of the series for {num_terms} terms is: {series_sum}")
```
